## Question 1

Write a program to find out how many counties are named `Knox` in the US.

```python
import ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت و مقداردهی اولیه Earth Engine
ee.Authenticate()
ee.Initialize(project=PROJECT_ID)

# بارگذاری مجموعه شهرستان‌های آمریکا و فیلتر کردن نام Knox
counties = ee.FeatureCollection('TIGER/2018/Counties')
knox_counties = counties.filter(ee.Filter.eq('NAME', 'Knox'))

# محاسبه و چاپ تعداد شهرستان‌های Knox
knox_count = knox_counties.size().getInfo()
print(f"Number of counties named 'Knox' in the US: {knox_count}")

# نمایش اسامی کامل برای بررسی دقیق‌تر
knox_info = knox_counties.aggregate_array('NAMELSAD').getInfo()
print("Counties found:", knox_info)

# ساخت نقشه با پایه بدون خطای 403 برای تهیه اسکرین‌شات
m = geemap.Map(center=[39.5, -93.5], zoom=4, height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')
m.addLayer(
    knox_counties.style(color='red', fillColor='ff000044', width=2),
    {},
    'Knox Counties',
)
m.centerObject(knox_counties, zoom=4)
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')

m
```

## Question 3

Use [Landsat-9](https://developers.google.com/earth-engine/datasets/catalog/LANDSAT_LC09_C02_T1_L2) data to create a cloud-free imagery for Knox County, TN. Display the imagery on the map with a proper band combination.

```python
import ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت (در صورت نیاز خط زیر را از حالت کامنت خارج کنید)
# ee.Authenticate()

# مقداردهی اولیه Earth Engine با شناسه پروژه ثبت‌شده
ee.Initialize(project=PROJECT_ID)

# بارگذاری مرز Knox County در ایالت Tennessee (STATEFP = '47')
counties = ee.FeatureCollection('TIGER/2018/Counties')
knox_tn = counties.filter(
    ee.Filter.And(ee.Filter.eq('NAME', 'Knox'), ee.Filter.eq('STATEFP', '47'))
)


# تابع نقاب‌گذاری ابر و سایه ابر در Landsat-9 Collection 2 Level 2 با استفاده از QA_PIXEL
def mask_l9_clouds(image):
  qa = image.select('QA_PIXEL')
  cloud_shadow_bit_mask = 1 << 4
  cloud_bit_mask = 1 << 3
  cirrus_bit_mask = 1 << 2
  dilated_cloud_bit_mask = 1 << 1

  mask = (
      qa.bitwiseAnd(dilated_cloud_bit_mask)
      .eq(0)
      .And(qa.bitwiseAnd(cirrus_bit_mask).eq(0))
      .And(qa.bitwiseAnd(cloud_bit_mask).eq(0))
      .And(qa.bitwiseAnd(cloud_shadow_bit_mask).eq(0))
  )

  optical_bands = image.select('SR_B.').multiply(0.0000275).add(-0.2)
  return image.addBands(optical_bands, overwrite=True).updateMask(mask)


# بارگذاری Landsat-9 C02 T1 L2، فیلتر محدوده، حذف ابر و محاسبه میانه (Median)
l9_composite = (
    ee.ImageCollection('LANDSAT/LC09/C02/T1_L2')
    .filterBounds(knox_tn)
    .filterDate('2022-01-01', '2024-12-31')
    .filter(ee.Filter.lt('CLOUD_COVER', 30))
    .map(mask_l9_clouds)
    .median()
    .clip(knox_tn)
)

# تنظیمات نمایش ترکیب رنگی کاذب (False Color: SR_B5, SR_B4, SR_B3)
vis_params = {
    'bands': ['SR_B5', 'SR_B4', 'SR_B3'],
    'min': 0.02,
    'max': 0.45,
    'gamma': 1.1,
}

# ساخت نقشه تعاملی با پایه CartoDB.Positron (بدون خطای 403 کاشی‌های OSM)
m = geemap.Map(height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# افزودن لایه تصویر بدون ابر Landsat-9 و حاشیه Knox County
m.addLayer(
    l9_composite, vis_params, 'Landsat-9 Cloud-Free False Color (Knox, TN)'
)
m.addLayer(
    knox_tn.style(color='yellow', fillColor='00000000', width=2),
    {},
    'Knox County, TN Border',
)

# مرکز قرار دادن نقشه روی Knox County
m.centerObject(knox_tn, zoom=9)

# افزودن متن نام و لوگو
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')
m.add_image('https://i.imgur.com/at4Qprk.png', position='bottomright')

# نمایش نقشه
m
```

## Question 4

Use [Sentinel-2](https://developers.google.com/earth-engine/datasets/catalog/COPERNICUS_S2_SR) data to create a cloud-free imagery for Knox County, TN. Display the imagery on the map with a proper band combination.

```python
# Add your code hereimport ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت (در صورت نیاز خط زیر را از حالت کامنت خارج کنید)
# ee.Authenticate()

# مقداردهی اولیه Earth Engine با شناسه پروژه ثبت‌شده
ee.Initialize(project=PROJECT_ID)

# بارگذاری مرز Knox County در ایالت Tennessee (STATEFP = '47')
counties = ee.FeatureCollection('TIGER/2018/Counties')
knox_tn = counties.filter(
    ee.Filter.And(ee.Filter.eq('NAME', 'Knox'), ee.Filter.eq('STATEFP', '47'))
)


# تابع نقاب‌گذاری ابر و سایه ابر با استفاده از باند SCL در Sentinel-2 SR
def mask_s2_clouds(image):
  scl = image.select('SCL')
  # حذف پیکسل‌های ابر متوسط (3)، ابر بالا (8)، ابر سیروس (9) و سایه ابر (10)
  clear_mask = (
      scl.neq(3).And(scl.neq(8)).And(scl.neq(9)).And(scl.neq(10))
  )
  return image.updateMask(clear_mask)


# بارگذاری Sentinel-2 SR، فیلتر محدوده و بازه زمانی، حذف ابر و محاسبه میانه (Median)
s2_composite = (
    ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
    .filterBounds(knox_tn)
    .filterDate('2023-01-01', '2024-12-31')
    .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 30))
    .map(mask_s2_clouds)
    .median()
    .clip(knox_tn)
)

# تنظیمات نمایش ترکیب رنگی کاذب (False Color: B8, B4, B3)
vis_params = {'bands': ['B8', 'B4', 'B3'], 'min': 200, 'max': 4000, 'gamma': 1.1}

# ساخت نقشه تعاملی با پایه CartoDB.Positron (بدون خطای 403 کاشی‌های OSM)
m = geemap.Map(height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# افزودن لایه تصویر بدون ابر Sentinel-2 و حاشیه Knox County
m.addLayer(
    s2_composite, vis_params, 'Sentinel-2 Cloud-Free False Color (Knox, TN)'
)
m.addLayer(
    knox_tn.style(color='yellow', fillColor='00000000', width=2),
    {},
    'Knox County, TN Border',
)

# مرکز قرار دادن نقشه روی Knox County
m.centerObject(knox_tn, zoom=9)

# افزودن متن نام و لوگو
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')
m.add_image('https://i.imgur.com/at4Qprk.png', position='bottomright')

# نمایش نقشه
m
```

## Question 5

Use [NAIP](https://developers.google.com/earth-engine/datasets/catalog/USDA_NAIP_DOQQ) imagery to create a cloud-free imagery for Knox County, TN. Display the imagery on the map with a proper band combination.

```python
import ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت (در صورت نیاز خط زیر را از حالت کامنت خارج کنید)
# ee.Authenticate()

# مقداردهی اولیه Earth Engine با شناسه پروژه ثبت‌شده
ee.Initialize(project=PROJECT_ID)

# بارگذاری مرز Knox County در ایالت Tennessee (STATEFP = '47')
counties = ee.FeatureCollection('TIGER/2018/Counties')
knox_tn = counties.filter(
    ee.Filter.And(ee.Filter.eq('NAME', 'Knox'), ee.Filter.eq('STATEFP', '47'))
)

# بارگذاری داده‌های NAIP، فیلتر محدوده و بازه زمانی، موزاییک کردن و برش روی Knox County
naip_knox = (
    ee.ImageCollection('USDA/NAIP/DOQQ')
    .filterBounds(knox_tn)
    .filterDate('2021-01-01', '2024-12-31')
    .mosaic()
    .clip(knox_tn)
)

# تنظیمات نمایش ترکیب رنگی کاذب فروسرخ نزدیک (False Color: N, R, G)
# (اگر در بازه زمانی انتخابی باند N موجود نبود، می‌توانید به ['R', 'G', 'B'] تغییر دهید)
vis_params = {'bands': ['N', 'R', 'G'], 'min': 0, 'max': 255}

# ساخت نقشه تعاملی با پایه CartoDB.Positron (بدون خطای 403 کاشی‌های OSM)
m = geemap.Map(height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# افزودن لایه NAIP و حاشیه Knox County
m.addLayer(naip_knox, vis_params, 'NAIP False Color (Knox, TN)')
m.addLayer(
    knox_tn.style(color='yellow', fillColor='00000000', width=2),
    {},
    'Knox County, TN Border',
)

# مرکز قرار دادن نقشه روی Knox County
m.centerObject(knox_tn, zoom=10)

# افزودن متن نام و لوگو
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')
m.add_image('https://i.imgur.com/at4Qprk.png', position='bottomright')

# نمایش نقشه
m
```
