## Question 1

Visualize the [USGS Watershed Boundary Dataset](https://developers.google.com/earth-engine/datasets/catalog/USGS_WBD_2017_HUC04) with outline color only, no fill color.

```python
import ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت و مقداردهی اولیه Earth Engine
ee.Authenticate()
ee.Initialize(project=PROJECT_ID)

# بارگذاری مجموعه داده مرز حوضه‌های آبریز USGS WBD (سطح HUC08)
wbd = ee.FeatureCollection('USGS/WBD/2017/HUC08')

# ساخت نقشه تعاملی با پایه CartoDB.Positron (بدون خطای 403 کاشی‌های OSM)
m = geemap.Map(center=[39.5, -98.35], zoom=4, height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# نمایش فقط حاشیه (outline only) بدون رنگ پرکننده (fillColor شفاف)
m.addLayer(
    wbd.style(color='blue', fillColor='00000000', width=1.5),
    {},
    'USGS WBD HUC08 (Outline Only)',
)

# افزودن متن نام و لوگو
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')
m.add_image('https://i.imgur.com/at4Qprk.png', position='bottomright')

# نمایش نقشه
m
```

## Question 2

Filter the USGS Watershed Boundary dataset and select the watershed that intersects the county of your choice.

```python
import ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت (در صورت نیاز خط زیر را از حالت کامنت خارج کنید)
# ee.Authenticate()

# مقداردهی اولیه Earth Engine با شناسه پروژه ثبت‌شده
ee.Initialize(project=PROJECT_ID)

# انتخاب شهرستان دلخواه: Knox County در Tennessee (STATEFP = '47')
counties = ee.FeatureCollection('TIGER/2018/Counties')
selected_county = counties.filter(
    ee.Filter.And(ee.Filter.eq('NAME', 'Knox'), ee.Filter.eq('STATEFP', '47'))
)

# فیلتر کردن حوضه‌های آبریز USGS WBD که با شهرستان منتخب تلاقی دارند
wbd = ee.FeatureCollection('USGS/WBD/2017/HUC08')
intersecting_watersheds = wbd.filterBounds(selected_county.geometry())

# نمایش تعداد و نام حوضه‌های آبریز منتخب
print(
    'Intersecting watersheds count:', intersecting_watersheds.size().getInfo()
)
print(
    'Watershed names:', intersecting_watersheds.aggregate_array('NAME').getInfo()
)

# ساخت نقشه تعاملی با پایه CartoDB.Positron (بدون خطای 403 کاشی‌های OSM)
m = geemap.Map(height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# نمایش حوضه‌های آبریز متقاطع با حاشیه آبی و مرز شهرستان با حاشیه قرمز
m.addLayer(
    intersecting_watersheds.style(color='blue', fillColor='1e90ff33', width=2),
    {},
    'Intersecting WBD Watersheds',
)
m.addLayer(
    selected_county.style(color='red', fillColor='00000000', width=2.5),
    {},
    'Knox County, TN',
)

# مرکز قرار دادن نقشه روی حوضه‌های منتخب
m.centerObject(intersecting_watersheds, zoom=9)

# افزودن متن نام و لوگو
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')
m.add_image('https://i.imgur.com/at4Qprk.png', position='bottomright')

# نمایش نقشه
m
```

## Question 3

Clip the [USGS 3DEP 10m DEM](https://developers.google.com/earth-engine/datasets/catalog/USGS_3DEP_10m) with the watershed that intersects the county of your choice. Display the DEM with a proper color palette and color bar.

```python
import ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت (در صورت نیاز خط زیر را از حالت کامنت خارج کنید)
# ee.Authenticate()

# مقداردهی اولیه Earth Engine با شناسه پروژه ثبت‌شده
ee.Initialize(project=PROJECT_ID)

# انتخاب شهرستان Knox County در Tennessee (STATEFP = '47')
counties = ee.FeatureCollection('TIGER/2018/Counties')
selected_county = counties.filter(
    ee.Filter.And(ee.Filter.eq('NAME', 'Knox'), ee.Filter.eq('STATEFP', '47'))
)

# فیلتر کردن حوضه‌های آبریز USGS WBD که با شهرستان منتخب تلاقی دارند
wbd = ee.FeatureCollection('USGS/WBD/2017/HUC08')
intersecting_watersheds = wbd.filterBounds(selected_county.geometry())
watershed_geom = intersecting_watersheds.geometry()

# بارگذاری USGS 3DEP 10m DEM به‌صورت ee.Image و برش با محدوده حوضه‌های آبریز منتخب
dem_3dep = ee.Image('USGS/3DEP/10m').clip(watershed_geom)

# تنظیمات پالت رنگی مناسب برای ارتفاع (متر)
vis_params = {
    'min': 200,
    'max': 1100,
    'palette': ['#2b83ba', '#abdda4', '#ffffbf', '#fdae61', '#d7191c'],
}

# ساخت نقشه تعاملی با پایه CartoDB.Positron (بدون خطای 403 کاشی‌های OSM)
m = geemap.Map(height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# افزودن لایه DEM برش‌خورده و مرز حوضه‌های آبریز
m.addLayer(
    dem_3dep, vis_params, 'USGS 3DEP 10m DEM (Intersecting Watersheds)'
)
m.addLayer(
    intersecting_watersheds.style(color='black', fillColor='00000000', width=2),
    {},
    'Intersecting Watersheds Border',
)

# افزودن نوار رنگ (Color Bar) به نقشه
m.add_colorbar(
    vis_params,
    label='Elevation (m)',
    layer_name='USGS 3DEP 10m DEM',
    position='bottomright',
)

# مرکز قرار دادن نقشه روی حوضه‌های آبریز منتخب
m.centerObject(intersecting_watersheds, zoom=9)

# افزودن متن نام و لوگو
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')
m.add_image('https://i.imgur.com/at4Qprk.png', position='bottomright')

# نمایش نقشه
m
```

## Question 4

Use the [USGS National Land Cover Database](https://developers.google.com/earth-engine/datasets/catalog/USGS_NLCD_RELEASES_2019_REL_NLCD) and [US Census States](https://developers.google.com/earth-engine/datasets/catalog/TIGER_2018_States) to create a split-panel map for visualizing land cover change (2001-2019) for a US state of your choice. Make sure you add the NLCD legend to the map.

```python
import ee
import geemap

PROJECT_ID = 'lab3-515123'

# احراز هویت (در صورت نیاز از حالت کامنت خارج کنید)
# ee.Authenticate()

# مقداردهی اولیه Earth Engine
ee.Initialize(project=PROJECT_ID)

# انتخاب ایالت Tennessee (STATEFP = '47') از لایه ایالت‌های سرشماری آمریکا
states = ee.FeatureCollection('TIGER/2018/States')
selected_state = states.filter(ee.Filter.eq('STATEFP', '47'))
state_geom = selected_state.geometry()

# بارگذاری داده‌های NLCD برای سال‌های 2001 و 2019 و برش روی ایالت Tennessee
nlcd_collection = ee.ImageCollection('USGS/NLCD_RELEASES/2019_REL/NLCD')

nlcd_2001 = (
    nlcd_collection.filter(ee.Filter.eq('system:index', '2001'))
    .first()
    .select('landcover')
    .clip(state_geom)
)

nlcd_2019 = (
    nlcd_collection.filter(ee.Filter.eq('system:index', '2019'))
    .first()
    .select('landcover')
    .clip(state_geom)
)

# ساخت لایه‌های Tile برای مقایسه دوپانله
left_layer = geemap.ee_tile_layer(nlcd_2001, {}, 'NLCD 2001')
right_layer = geemap.ee_tile_layer(nlcd_2019, {}, 'NLCD 2019')

# ساخت نقشه تعاملی با پایه CartoDB.Positron
m = geemap.Map(height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# فعال‌سازی نقشه دوپانله (2001 در چپ و 2019 در راست)
m.split_map(left_layer=left_layer, right_layer=right_layer)

# نمایش مرز ایالت به‌صورت فقط حاشیه
m.addLayer(
    selected_state.style(color='black', fillColor='00000000', width=2),
    {},
    'Tennessee Border',
)

# افزودن راهنمای NLCD
m.add_legend(title='NLCD Land Cover', builtin_legend='NLCD')

# مرکز قرار دادن نقشه روی ایالت Tennessee
m.centerObject(selected_state, zoom=7)

# افزودن امضا
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')

# نمایش نقشه
m
```

## Questions 5

Download OpenStreetMap data for a city of your choice and visualize the city boundary and restaurants in the city.

```python
!pip install -q osmnx
import ee
import geemap
import osmnx as ox

PROJECT_ID = 'lab3-515123'

# احراز هویت (در صورت نیاز از حالت کامنت خارج کنید)
# ee.Authenticate()

# مقداردهی اولیه Earth Engine
ee.Initialize(project=PROJECT_ID)

place = 'Knoxville, Tennessee'

# دریافت مرز اداری شهر از OpenStreetMap
city_boundary = geemap.osm_to_ee(place)

# دریافت رستوران‌ها (amenity='restaurant') با osmnx و تبدیل به Earth Engine FeatureCollection
gdf_restaurants = ox.features_from_place(place, tags={'amenity': 'restaurant'})
# تبدیل به مرکز هندسی (centroid) تا چندضلعی‌ها و نقاط یکدست و خوانا نمایش داده شوند
gdf_restaurants = gdf_restaurants.copy()
gdf_restaurants['geometry'] = gdf_restaurants.geometry.centroid
restaurants = geemap.geopandas_to_ee(gdf_restaurants[['geometry']])

# ساخت نقشه تعاملی با پایه CartoDB.Positron
m = geemap.Map(height='600px')
m.clear_layers()
m.add_basemap('CartoDB.Positron')

# نمایش مرز شهر (فقط حاشیه) و رستوران‌ها
m.addLayer(
    city_boundary.style(color='navy', fillColor='00000000', width=2.5),
    {},
    f'{place} Boundary',
)
m.addLayer(
    restaurants.style(color='red', pointSize=4, width=1.5),
    {},
    'Restaurants (OSM)',
)

# مرکز قرار دادن نقشه روی شهر انتخاب‌شده
m.centerObject(city_boundary, zoom=11)

# افزودن امضا
m.add_text('Made by Mahdieh Zakizadeh', fontsize=20, position='bottomright')

# نمایش نقشه
m
```
